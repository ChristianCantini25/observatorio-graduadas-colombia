```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("¡Proyecto de graduados iniciado!")
print("Versión de pandas:", pd.__version__)
```

```python
from pathlib import Path

# Ubicación de las bases de datos
ruta_datos = Path("../data/raw")

# Verificar los seis archivos
for anio in range(2020, 2026):
    archivo = ruta_datos / f"graduados_{anio}.xlsx"

    if archivo.exists():
        print(f"{anio}: Archivo encontrado")
    else:
        print(f"{anio}: Archivo NO encontrado")
```

```python
# Explorar la estructura de los archivos Excel

for anio in range(2020, 2026):
    archivo = ruta_datos / f"graduados_{anio}.xlsx"

    excel = pd.ExcelFile(archivo)

    print(f"\n{'=' * 45}")
    print(f"AÑO {anio}")
    print(f"Hojas disponibles: {excel.sheet_names}")
```

```python
# Revisar las primeras filas de cada archivo

hojas = {
    2020: "Sheet 1",
    2021: "Sheet 1",
    2022: "Graduados E.S. 2022",
    2023: "1.",
    2024: "1.",
    2025: "1."
}

for anio, hoja in hojas.items():

    archivo = ruta_datos / f"graduados_{anio}.xlsx"

    muestra = pd.read_excel(
        archivo,
        sheet_name=hoja,
        header=None,
        nrows=8
    )

    print(f"\n{'=' * 50}")
    print(f"AÑO {anio}")
    print(f"{'=' * 50}")

    display(muestra.iloc[:, :12])
```

```python
# Identificar la fila de encabezados de cada año

filas_encabezado = {}

for anio, hoja in hojas.items():

    archivo = ruta_datos / f"graduados_{anio}.xlsx"

    muestra = pd.read_excel(
        archivo,
        sheet_name=hoja,
        header=None,
        nrows=35
    )

    encontrado = False

    for indice, fila in muestra.iterrows():

        texto = " ".join(
            fila.dropna().astype(str).str.upper()
        )

        if (
            "INSTITUCIÓN DE EDUCACIÓN SUPERIOR" in texto
            and "SECTOR" in texto
        ):
            filas_encabezado[anio] = indice
            encontrado = True
            break

    if encontrado:
        print(
            f"{anio}: Encabezados en fila "
            f"{filas_encabezado[anio] + 1} de Excel"
        )
    else:
        print(f"⚠️ {anio}: Encabezados no identificados")
```

```python
# Identificar las columnas de cada base de datos

columnas_por_anio = {}

for anio, hoja in hojas.items():

    archivo = ruta_datos / f"graduados_{anio}.xlsx"

    muestra = pd.read_excel(
        archivo,
        sheet_name=hoja,
        header=filas_encabezado[anio],
        nrows=3
    )

    columnas_por_anio[anio] = list(muestra.columns)

    print(f"\n{'=' * 55}")
    print(f"AÑO {anio} - {len(muestra.columns)} COLUMNAS")
    print(f"{'=' * 55}")

    for numero, columna in enumerate(muestra.columns, start=1):
        print(f"{numero}. {columna}")
```

```python
# Consolidación de graduados 2020-2025

import unicodedata

def normalizar_texto(texto):
    texto = str(texto).strip().upper()
    texto = "".join(
        c for c in unicodedata.normalize("NFD", texto)
        if unicodedata.category(c) != "Mn"
    )
    return " ".join(texto.split())

# Variables que utilizaremos
variables = {
    "INSTITUCION DE EDUCACION SUPERIOR (IES)": "institucion",
    "SECTOR IES": "sector",
    "CARACTER IES": "caracter_institucion",
    "DEPARTAMENTO DE OFERTA DEL PROGRAMA": "departamento",
    "PROGRAMA ACADEMICO": "programa",
    "NIVEL ACADEMICO": "nivel_academico",
    "NIVEL DE FORMACION": "nivel_formacion",
    "AREA DE CONOCIMIENTO": "area_conocimiento",
    "SEXO": "sexo",
    "ANO": "anio",
    "SEMESTRE": "semestre",
    "GRADUADOS": "graduados"
}

bases = []

for anio, hoja in hojas.items():

    print(f"Procesando año {anio}...")

    archivo = ruta_datos / f"graduados_{anio}.xlsx"

    df = pd.read_excel(
        archivo,
        sheet_name=hoja,
        header=filas_encabezado[anio]
    )

    # Normalizar nombres de columnas
    df.columns = [normalizar_texto(c) for c in df.columns]

    # Verificar variables disponibles
    faltantes = set(variables) - set(df.columns)

    if faltantes:
        raise ValueError(
            f"En {anio} faltan columnas: {faltantes}"
        )

    # Seleccionar y renombrar
    df = df[list(variables)].rename(columns=variables)

    # Normalizar sexo
    df["sexo"] = df["sexo"].astype("string").str.strip().str.upper()

    df["sexo"] = df["sexo"].replace({
        "MUJER": "FEMENINO",
        "HOMBRE": "MASCULINO"
    })

    # Convertir conteos
    df["graduados"] = pd.to_numeric(
        df["graduados"], errors="coerce"
    )

    # Año de origen del archivo
    df["anio"] = anio

    # Excluir filas sin graduados válidos
    df = df.dropna(subset=["graduados"])

    bases.append(df)

    print(f"Registros cargados: {len(df):,}")

# Unir las seis bases
graduados = pd.concat(bases, ignore_index=True)

print("\nCONSOLIDACIÓN COMPLETADA")
print(f"Total de registros: {len(graduados):,}")
print(f"Total de columnas: {graduados.shape[1]}")

display(graduados.head())
```

```python
# PASO 12 - Validación inicial de calidad de datos

print("1. GRADUADOS POR AÑO")
resumen_anual = graduados.groupby("anio")["graduados"].agg(
    registros="count",
    total_graduados="sum"
)
display(resumen_anual)

print("\n2. CATEGORÍAS DE SEXO")
resumen_sexo = graduados.groupby(
    ["anio", "sexo"], dropna=False
)["graduados"].sum().unstack(fill_value=0)

display(resumen_sexo)

print("\n3. VALORES FALTANTES POR VARIABLE")
faltantes = graduados.isna().sum().to_frame(
    name="valores_faltantes"
)
faltantes["porcentaje"] = (
    faltantes["valores_faltantes"] / len(graduados) * 100
).round(2)

display(faltantes)

print("\n4. CONTROL DE CONTEOS")
print("Conteos negativos:",
      (graduados["graduados"] < 0).sum())

print("Conteos iguales a cero:",
      (graduados["graduados"] == 0).sum())

print("Conteos no enteros:",
      (graduados["graduados"] % 1 != 0).sum())
```

```python
# PASO 13 - Participación femenina por año

resumen = graduados.groupby(
    ["anio", "sexo"]
)["graduados"].sum().unstack(fill_value=0)

resumen["total"] = resumen.sum(axis=1)

resumen["participacion_femenina"] = (
    resumen["FEMENINO"] / resumen["total"] * 100
)

tabla_anual = resumen[
    ["FEMENINO", "MASCULINO", "total",
     "participacion_femenina"]
].copy()

tabla_anual["participacion_femenina"] = (
    tabla_anual["participacion_femenina"].round(2)
)

print("PARTICIPACIÓN FEMENINA POR AÑO")
display(tabla_anual)

# Visualización
plt.figure(figsize=(10, 5))

plt.plot(
    tabla_anual.index,
    tabla_anual["participacion_femenina"],
    marker="o",
    linewidth=2
)

plt.title("Participación femenina entre graduados (2020–2025)")
plt.xlabel("Año")
plt.ylabel("Participación femenina (%)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# PASO 14 - Participación femenina por área de conocimiento

# Agrupar por área y sexo
areas = graduados.groupby(
    ["area_conocimiento", "sexo"]
)["graduados"].sum().unstack(fill_value=0)

# Calcular el total de graduados
areas["total"] = areas.sum(axis=1)

# Calcular participación femenina
areas["participacion_femenina"] = (
    areas["FEMENINO"] / areas["total"] * 100
)

# Ordenar de mayor a menor
areas = areas.sort_values(
    "participacion_femenina",
    ascending=False
)

# Mostrar resultados
print("PARTICIPACIÓN FEMENINA POR ÁREA DE CONOCIMIENTO")

display(
    areas[
        ["FEMENINO", "MASCULINO",
         "total", "participacion_femenina"]
    ].round(2)
)

# Visualización
plt.figure(figsize=(11, 6))

plt.barh(
    areas.index,
    areas["participacion_femenina"]
)

plt.axvline(
    50,
    linestyle="--",
    color="gray",
    label="Paridad (50 %)"
)

plt.xlabel("Participación femenina (%)")
plt.ylabel("Área de conocimiento")
plt.title(
    "Participación femenina por área de conocimiento\n"
    "Colombia, 2020–2025"
)

plt.legend()
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# PASO 15 - Evolución de participación femenina por área

# Excluir categorías sin clasificación académica
datos_areas = graduados[
    ~graduados["area_conocimiento"].isin([
        "Sin clasificar",
        "Sin información"
    ])
].copy()

# Agrupar por año, área y sexo
evolucion = datos_areas.groupby(
    ["anio", "area_conocimiento", "sexo"]
)["graduados"].sum().unstack(fill_value=0)

# Calcular participación
evolucion["total"] = evolucion.sum(axis=1)

evolucion["participacion_femenina"] = (
    evolucion["FEMENINO"] / evolucion["total"] * 100
)

# Crear tabla comparativa
tabla_evolucion = evolucion[
    "participacion_femenina"
].unstack("anio").round(2)

print("EVOLUCIÓN POR ÁREA DE CONOCIMIENTO")
display(tabla_evolucion)

# Gráfico de evolución
plt.figure(figsize=(12, 7))

for area in tabla_evolucion.index:
    plt.plot(
        tabla_evolucion.columns,
        tabla_evolucion.loc[area],
        marker="o",
        label=area
    )

plt.axhline(50, linestyle="--", color="gray")
plt.title("Evolución de la participación femenina por área")
plt.xlabel("Año")
plt.ylabel("Participación femenina (%)")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# PASO 16 - Participación femenina por departamento

departamentos = graduados.groupby(
    ["departamento", "sexo"]
)["graduados"].sum().unstack(fill_value=0)

departamentos["total"] = departamentos.sum(axis=1)

departamentos["participacion_femenina"] = (
    departamentos["FEMENINO"] / departamentos["total"] * 100
)

# Ordenar por participación femenina
departamentos = departamentos.sort_values(
    "participacion_femenina",
    ascending=False
)

print("PARTICIPACIÓN FEMENINA POR DEPARTAMENTO")

display(
    departamentos[
        ["FEMENINO", "MASCULINO",
         "total", "participacion_femenina"]
    ].round(2)
)

# Gráfico comparativo
plt.figure(figsize=(11, 10))

plt.barh(
    departamentos.index,
    departamentos["participacion_femenina"]
)

plt.axvline(
    50,
    color="gray",
    linestyle="--",
    label="Paridad (50 %)"
)

plt.xlabel("Participación femenina (%)")
plt.ylabel("Departamento")
plt.title(
    "Participación femenina por departamento\n"
    "Colombia, 2020–2025"
)

plt.legend()
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# PASO 17 - Normalizar departamentos

import re
import unicodedata

def normalizar_departamento(valor):
    if pd.isna(valor):
        return "SIN INFORMACION"

    texto = str(valor).strip().upper()

    texto = "".join(
        c for c in unicodedata.normalize("NFD", texto)
        if unicodedata.category(c) != "Mn"
    )

    texto = re.sub(r"[^A-Z0-9 ]", " ", texto)
    texto = " ".join(texto.split())

    return texto

# Crear una nueva columna sin modificar los datos originales
graduados["departamento_normalizado"] = (
    graduados["departamento"].apply(normalizar_departamento)
)

# Revisar qué nombres originales quedaron agrupados
equivalencias = (
    graduados[
        ["departamento", "departamento_normalizado"]
    ]
    .drop_duplicates()
    .sort_values("departamento_normalizado")
)

print("EQUIVALENCIAS ENTRE DEPARTAMENTOS")
display(equivalencias)

print(
    "Departamentos originales:",
    graduados["departamento"].nunique()
)

print(
    "Departamentos normalizados:",
    graduados["departamento_normalizado"].nunique()
)
```

```python
# PASO 18 - Participación femenina por departamento corregida

departamentos_limpios = graduados.groupby(
    ["departamento_normalizado", "sexo"]
)["graduados"].sum().unstack(fill_value=0)

departamentos_limpios["total"] = departamentos_limpios.sum(axis=1)

departamentos_limpios["participacion_femenina"] = (
    departamentos_limpios["FEMENINO"]
    / departamentos_limpios["total"] * 100
)

departamentos_limpios = departamentos_limpios.sort_values(
    "participacion_femenina",
    ascending=False
)

print("PARTICIPACIÓN FEMENINA POR DEPARTAMENTO - DATOS LIMPIOS")

display(
    departamentos_limpios[
        ["FEMENINO", "MASCULINO",
         "total", "participacion_femenina"]
    ].round(2)
)

print(
    "\nNúmero de departamentos:",
    len(departamentos_limpios)
)

print(
    "\nTotal de graduados después de normalizar:",
    departamentos_limpios["total"].sum()
)

# Comprobar que los registros de San Andrés quedaron agrupados
print("\nREVISIÓN DE SAN ANDRÉS")

display(
    departamentos_limpios[
        departamentos_limpios.index.str.contains("SAN ANDRES")
    ][["total", "participacion_femenina"]].round(2)
)
```

```python
# PASO 19 - Participación femenina por nivel de formación

niveles = graduados.groupby(
    ["nivel_formacion", "sexo"]
)["graduados"].sum().unstack(fill_value=0)

# Calcular total de graduados
niveles["total"] = niveles.sum(axis=1)

# Calcular participación femenina
niveles["participacion_femenina"] = (
    niveles["FEMENINO"] / niveles["total"] * 100
)

# Ordenar resultados
niveles = niveles.sort_values(
    "participacion_femenina",
    ascending=False
)

print("PARTICIPACIÓN FEMENINA POR NIVEL DE FORMACIÓN")

display(
    niveles[
        ["FEMENINO", "MASCULINO",
         "total", "participacion_femenina"]
    ].round(2)
)

# Gráfico
plt.figure(figsize=(10, 6))

plt.barh(
    niveles.index,
    niveles["participacion_femenina"]
)

plt.axvline(
    50,
    color="gray",
    linestyle="--",
    label="Paridad (50 %)"
)

plt.xlabel("Participación femenina (%)")
plt.ylabel("Nivel de formación")

plt.title(
    "Participación femenina por nivel de formación\n"
    "Colombia, 2020–2025"
)

plt.legend()
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# PASO 20 - Normalización del nivel de formación

graduados["nivel_formacion_normalizado"] = (
    graduados["nivel_formacion"]
    .astype(str)
    .str.strip()
    .str.upper()
)

# Unificar denominaciones equivalentes
graduados["nivel_formacion_normalizado"] = (
    graduados["nivel_formacion_normalizado"].replace({
        "UNIVERSITARIA": "UNIVERSITARIO",
        "TECNOLÓGICA": "TECNOLÓGICO"
    })
)

# Revisar equivalencias
equivalencias_nivel = (
    graduados[
        ["nivel_formacion", "nivel_formacion_normalizado"]
    ]
    .drop_duplicates()
    .sort_values("nivel_formacion_normalizado")
)

print("EQUIVALENCIAS DE NIVELES DE FORMACIÓN")
display(equivalencias_nivel)

# Recalcular participación femenina
niveles_limpios = graduados.groupby(
    ["nivel_formacion_normalizado", "sexo"]
)["graduados"].sum().unstack(fill_value=0)

niveles_limpios["total"] = niveles_limpios.sum(axis=1)

niveles_limpios["participacion_femenina"] = (
    niveles_limpios["FEMENINO"]
    / niveles_limpios["total"] * 100
)

niveles_limpios = niveles_limpios.sort_values(
    "participacion_femenina",
    ascending=False
)

print("\nPARTICIPACIÓN FEMENINA POR NIVEL NORMALIZADO")

display(
    niveles_limpios[
        ["FEMENINO", "MASCULINO",
         "total", "participacion_femenina"]
    ].round(2)
)

print("\nCategorías originales:",
      graduados["nivel_formacion"].nunique())

print("Categorías normalizadas:",
      graduados["nivel_formacion_normalizado"].nunique())
```

```python
# PASO 21 - Participación femenina por características institucionales

for variable in ["sector", "caracter_institucion"]:

    tabla = graduados.groupby(
        [variable, "sexo"]
    )["graduados"].sum().unstack(fill_value=0)

    tabla["total"] = tabla.sum(axis=1)

    tabla["participacion_femenina"] = (
        tabla["FEMENINO"] / tabla["total"] * 100
    )

    tabla = tabla.sort_values(
        "participacion_femenina",
        ascending=False
    )

    print(f"\nPARTICIPACIÓN FEMENINA POR {variable.upper()}")

    display(
        tabla[
            ["FEMENINO", "MASCULINO",
             "total", "participacion_femenina"]
        ].round(2)
    )

    plt.figure(figsize=(10, 5))

    plt.barh(
        tabla.index,
        tabla["participacion_femenina"]
    )

    plt.axvline(
        50,
        color="gray",
        linestyle="--",
        label="Paridad (50 %)"
    )

    plt.title(
        f"Participación femenina por {variable}\n"
        "Colombia, 2020–2025"
    )

    plt.xlabel("Participación femenina (%)")
    plt.ylabel(variable.replace("_", " ").title())

    plt.legend()
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()
```

```python
# PASO 22 - Normalizar sector institucional

graduados["sector_normalizado"] = (
    graduados["sector"]
    .astype(str)
    .str.strip()
    .str.upper()
    .replace({
        "PRIVADA": "PRIVADO",
        "PRIVADO": "PRIVADO",
        "OFICIAL": "OFICIAL"
    })
)

# Calcular participación femenina por sector
sectores = graduados.groupby(
    ["sector_normalizado", "sexo"]
)["graduados"].sum().unstack(fill_value=0)

sectores["total"] = sectores.sum(axis=1)

sectores["participacion_femenina"] = (
    sectores["FEMENINO"] / sectores["total"] * 100
)

print("PARTICIPACIÓN FEMENINA POR SECTOR NORMALIZADO")

display(
    sectores[
        ["FEMENINO", "MASCULINO",
         "total", "participacion_femenina"]
    ].round(2)
)

print(
    "\nCategorías originales:",
    graduados["sector"].nunique()
)

print(
    "Categorías normalizadas:",
    graduados["sector_normalizado"].nunique()
)
```

```python
# PASO 23 - Control final de integridad

print("RESUMEN GENERAL")

print("Registros:", len(graduados))
print("Total de graduados:", int(graduados["graduados"].sum()))

print("\nTOTAL POR AÑO")
display(
    graduados.groupby("anio")["graduados"].sum()
)

print("\nTOTAL POR SEXO")
display(
    graduados.groupby("sexo")["graduados"].sum()
)

print("\nCONTROL DE DUPLICADOS")

# Las columnas originales que identifican una fila
columnas_originales = [
    "institucion",
    "sector",
    "caracter_institucion",
    "departamento",
    "programa",
    "nivel_academico",
    "nivel_formacion",
    "area_conocimiento",
    "sexo",
    "anio",
    "semestre",
    "graduados"
]

duplicados = graduados.duplicated(
    subset=columnas_originales,
    keep=False
)

print("Filas potencialmente duplicadas:", duplicados.sum())

if duplicados.any():
    display(
        graduados.loc[duplicados, columnas_originales]
        .head(10)
    )
```

```python
# PASO 24 - Investigar duplicados en la fuente original

archivo_2020 = ruta_datos / "graduados_2020.xlsx"

original_2020 = pd.read_excel(
    archivo_2020,
    sheet_name=hojas[2020],
    header=filas_encabezado[2020]
)

# Normalizar los encabezados
original_2020.columns = [
    normalizar_texto(c)
    for c in original_2020.columns
]

# Buscar el programa que apareció duplicado
ejemplo = original_2020[
    original_2020["PROGRAMA ACADEMICO"]
    .astype(str)
    .str.contains(
        "MAESTRÍA EN EDUCACIÓN POPULAR",
        case=False,
        na=False
    )
]

# Seleccionar registros de la Universidad del Cauca
ejemplo = ejemplo[
    ejemplo["INSTITUCION DE EDUCACION SUPERIOR (IES)"]
    .astype(str)
    .str.contains(
        "UNIVERSIDAD DEL CAUCA",
        case=False,
        na=False
    )
]

print("REGISTROS ORIGINALES DEL PROGRAMA")

display(ejemplo)
```

```python
# PASO 25 - Verificar duplicados exactos en la fuente

print("VALIDACIÓN DE DUPLICADOS - AÑO 2020")

# Duplicados considerando todas las columnas originales
duplicados_exactos = original_2020.duplicated(
    keep=False
)

print(
    "Filas duplicadas en las 39 columnas:",
    duplicados_exactos.sum()
)

print(
    "Total de filas originales:",
    len(original_2020)
)

# Mostrar ejemplos si existen
if duplicados_exactos.any():
    display(
        original_2020.loc[duplicados_exactos].head(10)
    )
else:
    print("No se encontraron duplicados exactos.")
```

```python
# PASO 26 - Revisión de duplicados exactos 2020-2025

resultados_duplicados = []

for anio, hoja in hojas.items():

    print(f"Revisando año {anio}...")

    if anio == 2020:
        df_original = original_2020.copy()
    else:
        archivo = ruta_datos / f"graduados_{anio}.xlsx"

        df_original = pd.read_excel(
            archivo,
            sheet_name=hoja,
            header=filas_encabezado[anio]
        )

    # Conservar solamente filas con conteos válidos
    df_original.columns = [
        normalizar_texto(c) for c in df_original.columns
    ]

    df_original["GRADUADOS"] = pd.to_numeric(
        df_original["GRADUADOS"], errors="coerce"
    )

    df_original = df_original.dropna(subset=["GRADUADOS"])

    # Comprobar duplicados considerando todas las columnas
    duplicados = df_original.duplicated(keep=False)

    resultados_duplicados.append({
        "anio": anio,
        "registros": len(df_original),
        "duplicados_exactos": int(duplicados.sum())
    })

# Resumen
tabla_duplicados = pd.DataFrame(resultados_duplicados)

display(tabla_duplicados)

print(
    "\nTotal de filas con duplicados exactos:",
    tabla_duplicados["duplicados_exactos"].sum()
)
```

```python
# PASO 27 - Exportar base procesada

from pathlib import Path

# Crear carpeta para los datos procesados
ruta_procesados = Path("../data/processed")
ruta_procesados.mkdir(parents=True, exist_ok=True)

# Guardar base consolidada
archivo_salida = ruta_procesados / "graduados_2020_2025.csv"

graduados.to_csv(
    archivo_salida,
    index=False,
    encoding="utf-8-sig"
)

print("BASE PROCESADA GUARDADA CORRECTAMENTE")
print("Ubicación:", archivo_salida.resolve())
print("Registros:", len(graduados))
print("Graduados:", int(graduados["graduados"].sum()))
print("Columnas:", len(graduados.columns))
```
